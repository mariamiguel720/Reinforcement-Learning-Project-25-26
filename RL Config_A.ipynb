# <font color='#2f94d7' size=6>**1. Project Overview**</font> <a class="anchor" id="1"></a>

[Back to TOC](#toc)

Configuration A applies **tabular reinforcement learning** to the ICU-Sepsis-v2 environment,
a discrete MDP derived from real MIMIC-III patient data. The state space consists of 716
discrete states (714 clinical plus 2 terminal: survived and died), and the action space
contains 25 treatment combinations (5 vasopressor levels x 5 IV fluid dose levels).

**Fixed configuration:** sofa_bias = 5.0, lam = 0.02 (as defined in env_setup.py).

**Implementation pattern (applied to every algorithm):**

1. Define a reusable training function `train_ALGO(hyperparameters, n_episodes, verbose)`
2. Optuna calls this function 50 times with different hyperparameter combinations (50k episodes each)
3. The best combination is used for the final training run (full episodes, verbose output)

**Primary evaluation metrics:**
- **Survival Rate:** policy evaluation over 1000 episodes after training
- **Learning Curve:** rolling 500-episode survival rate during training, showing the speed
  at which each algorithm learns to improve patient outcomes over time

policies and the baseline for the survival rate and learning curve comparisons.


```python
np.random.seed(SEED); env_r=make_sepsis_env(); rand_returns,rand_lengths=[],[]
for ep in range(1000):
    obs,_=env_r.reset(seed=np.random.randint(100_000)); total_r,steps,done=0.0,0,False
    while not done:
        obs,r,te,tr,_=env_r.step(env_r.action_space.sample()); total_r+=r; steps+=1; done=te or tr
    rand_returns.append(total_r); rand_lengths.append(steps)
env_r.close()
rand_returns=np.array(rand_returns); rand_lengths=np.array(rand_lengths)
rand_survival=float(np.mean(rand_returns>0))*100
rand_surv_ci =1.96*np.sqrt((rand_survival/100)*(1-rand_survival/100)/1000)*100
fig,axes=plt.subplots(1,3,figsize=(18,4))
fig.suptitle('3.4 Random Baseline',fontsize=13,fontweight='bold',color=H_COL)
axes[0].hist(rand_lengths,bins=20,color=B3,edgecolor='white',alpha=0.85)
axes[0].set_xlabel('Episode length (steps)'); axes[0].set_ylabel('Count')
axes[0].set_title('Episode Length Distribution',fontweight='bold')
axes[0].spines[['top','right']].set_visible(False)
axes[1].bar(['Survived','Died'],[rand_survival,100-rand_survival],
            color=[B2,'tomato'],edgecolor='white',alpha=0.9)
axes[1].set_ylabel('Episodes (%)'); axes[1].set_title('Outcome Distribution',fontweight='bold')
axes[1].spines[['top','right']].set_visible(False)
for i,v in enumerate([rand_survival,100-rand_survival]):
    axes[1].text(i,v+0.5,f'{v:.1f}%',ha='center',fontsize=11,fontweight='bold')
roll=smooth((rand_returns>0).astype(float)*100,50)
axes[2].plot(np.arange(50,1001),roll,color=REF_RAND,lw=1.8)
axes[2].axhline(rand_survival,color=B0,ls='--',lw=1.5,label=f'Mean={rand_survival:.1f}%')
axes[2].set_xlabel('Episode'); axes[2].set_ylabel('Survival Rate (%)')
axes[2].set_title('Rolling Survival Rate',fontweight='bold')
axes[2].legend(); axes[2].spines[['top','right']].set_visible(False)
plt.tight_layout(); plt.savefig(f'{PLOTS_DIR}/fig_3_4.png',dpi=150,bbox_inches='tight'); plt.show()
print(f'Random baseline: {rand_survival:.1f}% +/- {rand_surv_ci:.1f}%  |  Mean length: {np.mean(rand_lengths):.1f} steps')

rand_mean_ret = float(np.mean(rand_returns))
print(f'Random baseline: {rand_survival:.1f}% | mean return: {rand_mean_ret:.4f}')
```

## <font color='#2f94d7' size=6>**3.5 Feature-Outcome Relationship**</font> <a class="anchor" id="3_5"></a>

[Back to TOC](#toc)

Each of the 714 clinical states corresponds to a cluster centre in the 47-dimensional
physiological feature space shared with Config B. The correlation of each feature with
V_PI(s) (the survival probability under optimal treatment) reveals which physiological
variables are most associated with patient outcomes, providing a clinical interpretation
of the state space that bridges both configurations.


```python
env0=make_sepsis_env(); raw0=env0.unwrapped
P0=raw0._tx_mat.copy(); R0=(raw0._r_mat*raw0._tx_mat).sum(axis=2); env0.close()
pol0=np.zeros(S,dtype=int)
for _ in range(100):
    V0=np.zeros(S)
    for __ in range(5000):
        Vn=np.array([R0[s,pol0[s]]+GAMMA*P0[s,pol0[s]]@V0 for s in range(S)])
        if np.max(np.abs(Vn-V0))<1e-8: break
        V0=Vn
    new_pol0=np.argmax(R0+GAMMA*np.einsum('ijk,k->ij',P0,V0),axis=1)
    if np.array_equal(new_pol0,pol0): break
    pol0=new_pol0

V_preview=V0[:S-2]
feats=cluster_centers[:S-2]
corrs=np.array([np.corrcoef(feats[:,j],V_preview)[0,1] for j in range(47)])
sidx=np.argsort(np.abs(corrs))[::-1]
fig,axes=plt.subplots(1,2,figsize=(18,5))
fig.suptitle('3.5 Feature-Outcome Relationship',fontsize=13,fontweight='bold',color=H_COL)
top15=sidx[:15]
cols=[B2 if corrs[i]>=0 else 'tomato' for i in top15]
axes[0].barh([FEATURE_NAMES[i] for i in top15],corrs[top15],color=cols,edgecolor='white',alpha=0.9)
axes[0].axvline(0,color='grey',lw=1.0)
axes[0].set_xlabel('Pearson r with V_PI(s)')
axes[0].set_title('Top 15 Features Correlated with Survival Probability',fontweight='bold')
axes[0].spines[['top','right']].set_visible(False)
ca=[B2 if v>=0 else 'tomato' for v in corrs]
axes[1].bar(range(47),corrs,color=ca,edgecolor='none',alpha=0.8,width=0.8)
axes[1].set_xticks(range(47)); axes[1].set_xticklabels(FEATURE_NAMES,rotation=90,fontsize=7)
axes[1].axhline(0,color='grey',lw=0.8)
axes[1].set_ylabel('Pearson r with V_PI(s)'); axes[1].set_title('All 47 Feature Correlations',fontweight='bold')
axes[1].spines[['top','right']].set_visible(False)
plt.tight_layout(); plt.savefig(f'{PLOTS_DIR}/fig_3_5.png',dpi=150,bbox_inches='tight'); plt.show()
print('Top 5 positive:', [FEATURE_NAMES[i] for i in sidx[:5] if corrs[i]>0])
print('Top 5 negative:', [FEATURE_NAMES[i] for i in sidx if corrs[i]<0][:5])
```

```python
# REWARD SHAPING SETUP — Policy Iteration for V_PI computation
# It is computed here solely to obtain pi_V (the value function),
# P_pi (the transition matrix), and R_shaped (the shaped reward),

env_pi = fa.make_env(); raw_pi = env_pi.unwrapped
P_pi   = raw_pi._tx_mat.copy()
R_sa   = (raw_pi._r_mat * raw_pi._tx_mat).sum(axis=2)
env_pi.close()

pi_policy = np.zeros(S, dtype=int); V = np.zeros(S)
for it in range(1000):
    for _ in range(10000):
        Vn = np.array([R_sa[s, pi_policy[s]] + GAMMA * P_pi[s, pi_policy[s]] @ V
                       for s in range(S)])
        if np.max(np.abs(Vn - V)) < 1e-8: break
        V = Vn
    new_pol = np.argmax(R_sa + GAMMA * np.einsum('ijk,k->ij', P_pi, V), axis=1)
    if np.array_equal(new_pol, pi_policy): break
    pi_policy = new_pol

pi_V     = V.copy()
R_shaped = R_sa + GAMMA * np.einsum('ijk,k->ij', P_pi, pi_V) - pi_V[:, np.newaxis]

print(f'PI converged in {it+1} iteration(s)')
print(f'pi_V    : min={pi_V[:S-2].min():.4f}  mean={pi_V[:S-2].mean():.4f}  max={pi_V[:S-2].max():.4f}')
print(f'R_shaped: [{R_shaped.min():.4f}, {R_shaped.max():.4f}]')

# Evaluate PI policy to obtain pi_survival and pi_mean_ret
pi_eval, _  = fa.evaluate_policy(pi_policy, n_episodes=1000, label='Policy Iteration')
pi_survival = float(np.mean(pi_eval > 0)) * 100
pi_mean_ret = float(np.mean(pi_eval))
print(f'PI: {pi_survival:.1f}% | mean return: {pi_mean_ret:.4f}')

rand_mean_ret = float(np.mean(rand_returns))

# Inject MDP globals — required before any fa.train_* call
fa.configure(S, A, SEED, GAMMA, pi_V, P_pi, R_shaped)
```

## <font color='#2f94d7' size=6>**4.1 Rationale and Architecture**</font> <a class="anchor" id="5_1"></a>

[Back to TOC](#toc)

Dyna-Q (Sutton, 1990) extends Q-Learning by adding a model-based planning phase after each real environment step. After every real transition $(s, a, r, s')$, the algorithm performs $k$ additional simulated updates using a learned (or provided) model of the environment.

The standard Q-Learning update is:

$$Q(s, a) \leftarrow Q(s, a) + \alpha \left[ r + \gamma \max_{a'} Q(s', a') - Q(s, a) \right]$$

In Dyna-Q, this real update is followed by $k$ planning steps where a state-action pair $(s_p, a_p)$ is sampled from memory (or randomly), a next state is simulated from the model $P(s_n | s_p, a_p)$, and the same update rule is applied. This multiplies the effective number of Q-value updates per episode by $(1 + k)$ without requiring additional environment interaction.

**Prioritized Sweeping** (Moore & Atkeson, 1993) improves on random planning by maintaining a priority queue ordered by absolute TD error. After each real transition, if $|\delta| > \theta_{PS}$, the pair $(s, a)$ is added to the queue. The $k$ planning steps then pop the highest-priority pairs first, focusing computation where the Q-table is most uncertain. When the queue empties, remaining planning steps fall back to random sampling over clinical states.

**Reward shaping** (Ng et al., 1999) is applied to both real and planning updates:

$$r'(s, a, s') = r(s, a, s') + \gamma \cdot V_\pi(s') \cdot (1 - \text{done}) - V_\pi(s)$$

where $V_\pi$ is the Policy Iteration value function. This transformation is provably policy-invariant: the optimal policy under $r'$ is identical to the optimal policy under $r$, but the dense per-step signal accelerates learning by providing immediate feedback on whether an action moved the patient towards or away from survival.

### Clinical context

In the ICU sepsis setting, Dyna-Q+PS has a distinctive theoretical advantage: it can extract maximum information from each patient encounter by replaying its model-based knowledge. Each real episode generates one real transition but $k$ additional simulated Q-value updates, dramatically improving sample efficiency.

**Advantages in a real clinical setting:**
- **Sample efficiency.** Patient encounters in ICU are expensive (in time, ethics and outcome variance). An algorithm that learns more per episode is preferred for offline RL on historical data such as MIMIC-III, where the total number of trajectories is bounded.
- **Replay of rare patient states.** The Prioritized Sweeping mechanism revisits state-action pairs with high TD error, which often correspond to rare or critical patient situations. This is clinically relevant because rare cases (e.g., very high SOFA scores) are precisely where standard model-free methods struggle to learn.
- **Theoretical link to clinical decision support.** With access to a transition model, the agent's planning step resembles clinical mental simulation: "if I administer this treatment, what is the likely next state?"

**Disadvantages and risks:**
- **Model bias.** Dyna-Q requires an accurate transition model $P(s'|s,a)$. In a real clinical deployment, the true patient dynamics may differ from the model trained on historical data (distribution shift). A biased model could propagate errors through $k$ planning steps and lead to systematically suboptimal recommendations.
- **Computational cost.** With $k$ planning steps per real step, each training episode costs $(k+1) \times$ the wall-clock time of plain Q-Learning. For $k=95$ (the value found by Optuna), this is a 96x increase in compute per episode.
- **Black-box planning.** The planning phase modifies Q-values based on simulated transitions that never occurred, which can be difficult to audit for clinical safety.

### Hyperparameters

- $\alpha_{start}, \alpha_{end}, \alpha_{decay}$: learning rate schedule $\alpha_t = \max(\alpha_{end}, \alpha_{start} \cdot \alpha_{decay}^{\,t})$.
- $k$: number of planning steps per real environment step.
- $\theta_{PS}$: threshold below which TD errors are not added to the priority queue.
- $\varepsilon_{decay}$: exploration schedule $\varepsilon_t = \max(0.005, \varepsilon_{decay}^{\,t})$.

## <font color='#2f94d7' size=6>**5.1 Rationale and Architecture**</font> <a class="anchor" id="6_1"></a>

[Back to TOC](#toc)

**Why Q-Learning?**

Q-Learning is the foundational off-policy TD algorithm and the most widely studied
benchmark in tabular RL. Its off-policy nature (learning the optimal Q-values
regardless of the exploration policy) makes it sample-efficient and directly
comparable to Dyna-Q+PS, which uses Q-Learning updates internally.

**Learning rate decay** (alpha_start to alpha_end over training) is essential to prevent
late-training instability: a high initial alpha enables fast early learning while the
decay ensures Q-values converge to stable estimates as training progresses.

**Key comparison with Dyna-Q+PS:** Q-Learning has one Q-value update per environment
step. At equal episode budgets (60k), the difference in performance directly quantifies
the value of model-based planning in the ICU-Sepsis environment.

**Hyperparameters optimised by Optuna:** alpha_start, alpha_end, alpha_decay, eps_decay.


Q-Learning (Watkins, 1989) is the canonical off-policy Temporal Difference algorithm for control. It learns the optimal action-value function $Q^*(s, a)$ directly from sampled transitions, regardless of the policy used to generate those transitions.

The Q-Learning update rule is:

$$Q(s, a) \leftarrow Q(s, a) + \alpha \left[ r + \gamma \max_{a'} Q(s', a') - Q(s, a) \right]$$

The term $\max_{a'} Q(s', a')$ is what makes Q-Learning **off-policy**: it uses the greedy action at $s'$ for the update, even if the actual next action taken (under $\varepsilon$-greedy exploration) was different. This separation between behaviour policy and target policy is theoretically valuable because it allows learning from any source of experience, including historical clinician decisions.

Under tabular conditions with bounded rewards and a Robbins-Monro learning rate schedule, Q-Learning is guaranteed to converge to $Q^*$ in the limit (Watkins & Dayan, 1992).

We apply the same potential-based reward shaping as for Dyna-Q+PS:

$$r'(s, a, s') = r + \gamma V_\pi(s') (1 - \text{done}) - V_\pi(s)$$

The greedy policy at evaluation time is:

$$\pi(s) = \arg\max_a Q(s, a)$$

with $\varepsilon$-greedy exploration during training:

$$P(a) = \begin{cases} 1 - \varepsilon + \varepsilon / |A| & \text{if } a = \arg\max_a Q(s,a) \\ \varepsilon / |A| & \text{otherwise} \end{cases}$$

### Clinical context

Q-Learning is the natural baseline for any RL clinical application because it makes minimal assumptions about the underlying decision process.

**Advantages in a real clinical setting:**
- **No model required.** Q-Learning learns purely from observed transitions, making it directly applicable to historical clinical data without requiring an explicit transition model. This is critical when patient dynamics are too complex to model accurately.
- **Off-policy learning.** Q-Learning can learn optimal policies from data generated by suboptimal behaviour policies, including historical clinician decisions. This is essential for offline RL on retrospective datasets like MIMIC-III.
- **Interpretability.** The resulting Q-table maps each state-action pair to an explicit value, which can be inspected and audited.
- **Theoretical guarantees.** Convergence to $Q^*$ is proven under tabular conditions.

**Disadvantages and risks:**
- **Sample inefficiency.** Each real environment step produces only one Q-value update. In a 716-state, 25-action MDP (17,900 Q-table entries), learning every entry requires many episodes. Without planning, Q-Learning converges substantially slower than Dyna-Q+PS at equal episode budgets.
- **Maximisation bias.** The $\max_{a'} Q(s', a')$ term overestimates Q-values when there is noise in estimates. In sparse-reward clinical settings, this can lead to optimistic policies that recommend treatments based on insufficient evidence.
- **Sensitivity to learning rate schedule.** If $\alpha$ decays too quickly to a low floor, the Q-table effectively freezes and further training brings no improvement.

### Hyperparameters

- $\alpha_{start}, \alpha_{end}, \alpha_{decay}$: same schedule as Dyna-Q+PS.
- $\varepsilon_{decay}$: exploration schedule.
- Notably, Q-Learning has no analogue to Dyna-Q's $k$ planning parameter. The number of Q-value updates per real episode is always 1.

# <font color='#2f94d7' size=6>**8. Creative Extension: Clinical Decision Support Analysis**</font> <a class="anchor" id="9"></a>

[Back to TOC](#toc)

This section moves beyond algorithm evaluation metrics to ask the question that matters
most in a clinical setting: 
**What would a physician learn by observing this agent, and do its decisions make clinical sense?**

Three analyses use SHAP (SHapley Additive exPlanations) and policy analysis to interpret
the agents' learned behaviour in terms of the 47 physiological variables. Each analysis is designed to answer a concrete clinical question.

Standard RL evaluation focuses on outcome metrics — survival rate, return, agreement with the optimal policy. These tell us *how well* an agent performs but not *why* it makes specific decisions or *what clinical concepts* it learned. For a clinical decision support system, the latter questions are arguably more important: a physician needs to understand and trust the agent's reasoning before adopting its recommendations.

This extension builds on the two featured algorithms (Dyna-Q+PS and Q-Learning) and the V_PI reference policy to ask three concrete clinical interpretability questions:

| Clinical question | Method |
|-------------------|--------|
|  Has the agent learned a meaningful measure of patient severity that aligns with established clinical scores? | Surrogate SHAP on V_PI |
|  Does the optimal policy escalate treatment for sicker patients, or does it follow a different clinical logic? | Treatment intensity per V_PI quartile |
|  When two algorithms reach different policies, do they also weight clinical features differently? | SHAP correlation across policies |

The novelty lies in applying explainable-AI techniques (SHAP) that are standard in supervised learning to **value functions** in tabular RL. Q-tables do not operate directly on the 47 physiological features (they index discrete states), so SHAP cannot be applied directly. We bridge this gap with **surrogate models**: each algorithm's value function is fitted by a GradientBoosting regressor on the 47 features, and SHAP is applied to that surrogate. This is an established but uncommon technique in RL interpretability literature.

The 714 clinical states are derived from k-means clustering of patient feature vectors. Each state corresponds to a centroid in the 47-dimensional feature space. ### Clinical question

Two algorithms can disagree on the recommended treatment for the same patient while *also* disagreeing on which physiological variables matter most. We separate these two notions: **policy agreement** (do they choose the same action?) and **feature importance similarity** (do they weight the same variables?).

### Methodology

1. For each algorithm, fit a GradientBoostingRegressor to predict $\max_a Q(s, a)$ from the 47 features.
2. Compute SHAP values for each surrogate.
3. Compute the Pearson correlation between each algorithm's normalised SHAP feature importance and V_PI's feature importance (from 8.1).
4. Higher correlation = algorithm learned to weight features similarly to the theoretically optimal policy.

We expected Dyna-Q+PS (90.5% policy agreement with PI) to also show higher feature importance correlation with V_PI than Q-Learning (33.1% agreement). The intuition is that an algorithm that agrees on actions should also agree on which features drive those actions.
